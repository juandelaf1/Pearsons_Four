#### Outliers IQR + z-score

```python
sal = df_clean['normalized_salary'].dropna()
# IQR
Q1, Q3 = sal.quantile(0.25), sal.quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers_iqr = sal[(sal < lower) | (sal > upper)]
print(f"IQR limits: ${lower:,.0f} - ${upper:,.0f}")
print(f"IQR outliers: {len(outliers_iqr)} ({len(outliers_iqr)/len(sal)*100:.1f}%)")
# Z-score
z = np.abs(stats.zscore(sal))
outliers_z = sal[z > 3]
print(f"Z-score outliers: {len(outliers_z)} ({len(outliers_z)/len(sal)*100:.1f}%)")
# Suspicious values
print(f"\nSalary < $1,000: {len(sal[sal < 1000])} rows (likely errors)")
print(f"Salary > $1,000,000: {len(sal[sal > 1000000])} rows (likely outliers)")
```

#### Numeric Correlation Matrix

```python
# 4. Correlation Matrix for Numeric Variables
# We select only numeric columns to avoid errors
numeric_cols = df_stats[['normalized_salary', 'views', 'applies', 'experience_level_num', 'remote_allowed']].dropna()
correlation_matrix = numeric_cols.corr(method='pearson')

print("🔗 Correlation Matrix (Numeric Variables):")
correlation_matrix['normalized_salary'].sort_values(ascending=False)
```

```python
# === SPEARMAN CORRELATION (robust to skewness) ===
from scipy.stats import spearmanr

spearman_matrix = numeric_cols.corr(method='spearman')
print('=== Spearman Correlations with Salary ===')
print(spearman_matrix['normalized_salary'].sort_values(ascending=False))

# Compare Pearson vs Spearman for key pairs
print('\n=== Pearson vs Spearman Comparison ===')
for col in ['views', 'applies', 'experience_level_num', 'remote_allowed']:
    if col in numeric_cols.columns:
        r_p, p_p = pearsonr(numeric_cols['normalized_salary'], numeric_cols[col])
        r_s, p_s = spearmanr(numeric_cols['normalized_salary'], numeric_cols[col])
        print(f'{col:25s} | Pearson r={r_p:.3f} (p={p_p:.4f}) | Spearman rho={r_s:.3f} (p={p_s:.4f})')

print('\n---')
print('INTERPRETACION: Spearman es mas robusto que Pearson cuando la distribucion')
print('tiene asimetria positiva (como el salario). Si rho > r, la relacion es monotonica')
print('pero no lineal, lo que refuerza el uso de la mediana como medida central.')
```

#### Hypothesis Testing (Experience vs. Salary)

```python
# 9. One-way ANOVA test
# Comparing means across different experience levels
groups = [group['normalized_salary'].values for name, group in df_clean.groupby('formatted_experience_level')]
f_stat, p_value = stats.f_oneway(*groups)

print("🧪 Hypothesis Test (ANOVA):")
print(f"F-Statistic: {f_stat:.2f}")
print(f"P-Value: {p_value:.4f}")

if p_value < 0.05:
    print("Conclusion: Significant difference found between experience levels (Reject H0).")
else:
    print("Conclusion: No significant difference found (Fail to reject H0).")
```

```python
# === BOOTSTRAP CONFIDENCE INTERVAL FOR MEDIAN ===
np.random.seed(42)
sal_data = df_stats['normalized_salary'].dropna().values

n_bootstrap = 10000
bootstrap_medians = np.zeros(n_bootstrap)
for b in range(n_bootstrap):
    sample = np.random.choice(sal_data, size=len(sal_data), replace=True)
    bootstrap_medians[b] = np.median(sample)

ci_lower = np.percentile(bootstrap_medians, 2.5)
ci_upper = np.percentile(bootstrap_medians, 97.5)

print(f'Median salary: ${np.median(sal_data):,.0f}')
print(f'95% CI (bootstrap): [${ci_lower:,.0f} - ${ci_upper:,.0f}]')
print(f'Margin of error: +-${(ci_upper - ci_lower)/2:,.0f}')
print('\nThe bootstrap CI does not assume normality, making it')
print('appropriate for skewed salary distributions.')
```

```python
# === T-TEST: REMOTE PREMIUM SIGNIFICANCE ===
from scipy.stats import ttest_ind

remote_sal = df_stats[df_stats['is_remote']]['normalized_salary'].dropna()
onsite_sal = df_stats[~df_stats['is_remote']]['normalized_salary'].dropna()

t_stat, p_val = ttest_ind(remote_sal, onsite_sal, equal_var=False)

print(f'Remote count: {len(remote_sal)}, On-site count: {len(onsite_sal)}')
print(f'Remote mean: ${remote_sal.mean():,.0f}, On-site mean: ${onsite_sal.mean():,.0f}')
print(f'Remote median: ${np.median(remote_sal):,.0f}, On-site median: ${np.median(onsite_sal):,.0f}')
print(f'Premium: +{((np.median(remote_sal)/np.median(onsite_sal))-1)*100:.1f}%')
print(f'T-test (Welch): t={t_stat:.2f}, p={p_val:.6f}')
if p_val < 0.001:
    print('Result: Highly significant (p < 0.001) - Remote premium is statistically validated.')
elif p_val < 0.05:
    print('Result: Significant (p < 0.05).')
else:
    print('Result: Not significant at alpha=0.05.')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the clean data
df_viz = pd.read_csv('linkedin_data_roles_raw_nulls.csv')

df_sal = df_viz.dropna(subset=['normalized_salary'])

sns.set_theme(style="whitegrid")


plt.figure(figsize=(12, 6))

exp_order = ['Entry level', 'Associate', 'Mid-Senior level', 'Director', 'Executive']
sns.boxplot(data=df_sal, x='formatted_experience_level', y='normalized_salary', order=exp_order, palette='Set2')
plt.title('Salary Spread by Experience Level', fontsize=14, fontweight='bold')
plt.xlabel('Experience Level', fontsize=12)
plt.ylabel('Normalized Yearly Salary (USD)', fontsize=12)
plt.ylim(0, 500000)
plt.show()

```

> El boxplot permite visualizar simultaneamente mediana, IQR y outliers por nivel de experiencia.  
> Se prefirio sobre barras porque muestra la distribucion completa, no solo una medida central.  
> La progresion salarial es claramente positiva: de Entry (~$84K) a Executive (~$200K).  
> El nivel Internship se excluyo por tener datos anomalos (media de $963K, probablemente errores de codificacion).

#### Distribution of Data Role Salaries (with KDE) - Histogram

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df_sal, x='normalized_salary', kde=True, bins=40, color='teal')
plt.title('Distribution of Data Role Salaries (with KDE)', fontsize=14, fontweight='bold')
plt.xlabel('Normalized Yearly Salary (USD)', fontsize=12)
plt.ylabel('Number of Job Postings', fontsize=12)
plt.xlim(0, 400000)
plt.show()
```

> **Grafico 2 - Pie de pagina - Histograma + KDE:**  
> El histograma con curva KDE muestra la forma de la distribucion salarial.  
> La asimetria positiva (skewness > 0) es evidente: cola derecha hacia salarios altos.  
> La media ($96,795) es mayor que la mediana ($82,500), lo que justifica usar la mediana como medida central.  
> Elegimos KDE en lugar de curva normal teorica porque no asume normalidad.

#### Correlation Heatmap: Salary, Views, Applies, and Experience - Heatmap

```python
df_encoded = pd.get_dummies(df_viz, columns=['formatted_experience_level'])

exp_columns = [col for col in df_encoded.columns if 'formatted_experience_level_' in col]

columns_for_math = ['normalized_salary', 'views', 'applies'] + exp_columns

corr_matrix = df_encoded[columns_for_math].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap: Salary, Views, Applies, and Specific Experience Levels', fontsize=14, fontweight='bold')
plt.show()
```

> **Grafico 3 - Pie de pagina - Heatmap de correlacion:**  
> Se uso el coeficiente de Pearson por su interpretabilidad directa (-1 a +1).  
> La experiencia (Mid-Senior, Director, Executive) correlaciona positivamente con salario.  
> Views y Applies muestran correlacion moderada entre si (r ~0.62).  
> Advertencia: correlacion no implica causalidad. Ver Spearman como validacion.
